# 🎁 เพิ่ม Promotion Class ใน Class Diagram

## เหตุผลที่ต้องเพิ่ม Promotion

ระบบปัจจุบันมี **Campaign** (แคมเปญการตลาด) แต่ยังขาด **Promotion** (ระบบส่วนลด/โปรโมชั่น) ที่ใช้งานกับ Booking โดยตรง

### ความแตกต่าง Campaign vs Promotion

| Aspect | Campaign | Promotion |
|--------|----------|-----------|
| **วัตถุประสงค์** | การตลาด, สร้าง Brand Awareness | ส่วนลด, กระตุ้นยอดขาย |
| **ผู้รับผิดชอบ** | MarketingStaff | MarketingStaff |
| **ใช้โดย** | ทีมการตลาด (วิเคราะห์ Performance) | Customer (ใช้โค้ดส่วนลด) |
| **เชื่อมกับ** | MarketingStaff | **Booking** ✨ (ใช้จริง) |
| **ตัวอย่าง** | "Summer Marketing Campaign 2025" | "SUMMER20" (20% off) |

---

## 📋 Promotion Class Specification

### Attributes
- `promotionId : int {PK}` - รหัสโปรโมชั่น
- `campaignId : int {FK}` - เชื่อมกับ Campaign (nullable)
- `promotionCode : String(20) {UK}` - รหัสโปรโมชั่น เช่น "NEWUSER50"
- `promotionName : String(100)` - ชื่อโปรโมชั่น
- `promotionType : String(30)` - ประเภท: PERCENTAGE, FIXED_AMOUNT, FREE_SERVICE
- `discountValue : Decimal(8,2)` - มูลค่าส่วนลด (% หรือ บาท)
- `minBookingAmount : Decimal(10,2)` - ยอดขั้นต่ำที่ใช้ได้
- `maxDiscountAmount : Decimal(10,2)` - ส่วนลดสูงสุด (สำหรับ %)
- `usageLimit : int` - จำนวนครั้งที่ใช้ได้ทั้งหมด
- `usedCount : int` - จำนวนครั้งที่ใช้ไปแล้ว
- `customerUsageLimit : int` - ลูกค้า 1 คนใช้ได้กี่ครั้ง
- `startDate : Date` - วันเริ่มต้น
- `endDate : Date` - วันสิ้นสุด
- `isActive : boolean` - สถานะเปิด/ปิด
- `description : Text(500)` - รายละเอียด
- `createdBy : int {FK}` - Employee ที่สร้าง

### Methods
- `+ validateCode() : boolean` - ตรวจสอบความถูกต้องของโค้ด
- `+ isAvailable() : boolean` - ตรวจสอบว่าใช้งานได้หรือไม่
- `+ calculateDiscount(amount) : Decimal` - คำนวณส่วนลด
- `+ applyToBooking(booking) : boolean` - ใช้กับ Booking
- `+ incrementUsage() : void` - เพิ่มจำนวนการใช้งาน
- `+ checkCustomerEligibility(customerId) : boolean` - ตรวจสอบสิทธิ์ลูกค้า
- `+ getRemainingUsage() : int` - ดูจำนวนครั้งคงเหลือ

## 🔗 Relationships ที่ต้องเพิ่ม

1. **Promotion (0..1) ← applies ← (1) Booking**
   - Booking สามารถใช้ Promotion ได้ 0 หรือ 1 รายการ
   - Promotion สามารถถูกใช้กับ Booking หลายรายการ

2. **Campaign (1) → creates → (0..*) Promotion**
   - Campaign 1 รายการสามารถสร้าง Promotion ได้หลายรายการ (optional)
   - Promotion อาจเป็นอิสระไม่เชื่อมกับ Campaign ก็ได้

3. **MarketingStaff (1) → manages → (0..*) Promotion**
   - MarketingStaff จัดการ Promotion

## 📊 Database Schema

```sql
CREATE TABLE promotions (
    promotion_id INT AUTO_INCREMENT PRIMARY KEY,
    campaign_id INT NULL,
    promotion_code VARCHAR(20) UNIQUE NOT NULL,
    promotion_name VARCHAR(100) NOT NULL,
    promotion_type VARCHAR(30) NOT NULL,
    discount_value DECIMAL(8,2) NOT NULL,
    min_booking_amount DECIMAL(10,2) DEFAULT 0,
    max_discount_amount DECIMAL(10,2),
    usage_limit INT DEFAULT 0,
    used_count INT DEFAULT 0,
    customer_usage_limit INT DEFAULT 1,
    start_date DATE NOT NULL,
    end_date DATE NOT NULL,
    is_active BOOLEAN DEFAULT TRUE,
    description TEXT,
    created_by INT NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP ON UPDATE CURRENT_TIMESTAMP,
    
    FOREIGN KEY (campaign_id) REFERENCES campaigns(campaign_id) ON DELETE SET NULL,
    FOREIGN KEY (created_by) REFERENCES employees(employee_id),
    
    INDEX idx_promotion_code (promotion_code),
    INDEX idx_dates (start_date, end_date),
    INDEX idx_active (is_active)
) ENGINE=InnoDB;
```

```java
public Decimal calculateDiscount(Decimal bookingAmount) {
    if (bookingAmount < this.minBookingAmount) {
        return 0.0;
    }
    
    Decimal discount = 0.0;
    
    switch (this.promotionType) {
        case "PERCENTAGE":
            discount = bookingAmount * (this.discountValue / 100);
            if (this.maxDiscountAmount != null && discount > this.maxDiscountAmount) {
                discount = this.maxDiscountAmount;
            }
            break;
        case "FIXED_AMOUNT":
            discount = this.discountValue;
            break;
        case "FREE_SERVICE":
            // Logic for free service
            break;
    }
    
    return discount;
}
```

# 3.2.6 Persistent Data Design

## Database Schema Specification

รายละเอียดโครงสร้างฐานข้อมูลทั้งหมดในระบบ SportsBooking พร้อม Attributes, Keys, Data Types และ Descriptions

---

## 3.2.6.1 Account

| Attribute | Key | Type | Size | Description |
|-----------|-----|------|------|-------------|
| account_id | PK | int | 11 | รหัสบัญชีผู้ใช้ (Auto Increment) |
| email | FK | varchar | 100 | อีเมลบุคคล (อ้างอิง Person) |
| username | UK | varchar | 50 | ชื่อผู้ใช้สำหรับ Login |
| password | | varchar | 255 | รหัสผ่าน (เข้ารหัส bcrypt/hash) |
| last_login | | timestamp | - | เวลา Login ครั้งล่าสุด |
| is_active | | boolean | 1 | สถานะเปิดใช้งาน (1=active, 0=inactive) |
| failed_login_attempts | | int | 3 | จำนวนครั้งที่ Login ผิดพลาด |
| created_at | | timestamp | - | วันที่สร้างบัญชี |
| updated_at | | timestamp | - | วันที่อัพเดทล่าสุด |

**Indexes:**
- PRIMARY KEY (account_id)
- UNIQUE KEY (username)
- FOREIGN KEY (email) REFERENCES persons(email) ON DELETE CASCADE
- INDEX idx_email (email)

| Attribute | Key | Type | Size | Description |
|-----------|-----|------|------|-------------|
| email | PK | varchar | 100 | อีเมลสำหรับ Login และติดต่อ (Primary Key) |
| first_name | | varchar | 100 | ชื่อ |

**Indexes:**
- PRIMARY KEY (email)
- INDEX idx_name (first_name, last_name)

**หมายเหตุ:** เปลี่ยนจาก `person_id` เป็น `email` เป็น Primary Key ตามคำแนะนำของอาจารย์ เนื่องจากผู้ใช้จำ email ได้ง่ายกว่า และใช้สำหรับยืนยันตัวตน/รีเซ็ตรหัสผ่าน
