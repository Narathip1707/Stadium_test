## 🔍 **2.2 Validation Logic (จุดเด่นของระบบ)** [1.5 นาที]

### **Speaking Points:**
```
จุดเด่นของ Use Case Description ของเรา
คือมีการ Validate ทุกขั้นตอนอย่างละเอียด
และมีการจัดการทั้ง Success และ Failure Scenarios
```

### **Example: Register User Validation**

#### **Step 1.1.4 - System Validation:**
```
ระบบตรวจสอบความถูกต้องของข้อมูล:
  ✓ ตรวจสอบรูปแบบอีเมล (must contain @)
  ✓ ตรวจสอบอีเมลซ้ำในฐานข้อมูล
  ✓ ตรวจสอบความยาวรหัสผ่าน (≥ 8 ตัวอักษร)
  ✓ ตรวจสอบความแข็งแรงของรหัสผ่าน
  ✓ ตรวจสอบเบอร์โทรศัพท์ (10 หลัก)
  ✓ ตรวจสอบชื่อ-นามสกุล (ไม่ว่าง)
```

#### **Step 1.1.5 - Binary Outcomes:**

**✅ Success Path (1.1.5.1):**
```
- ระบบบันทึกข้อมูลลงฐานข้อมูล
- ระบบสร้าง Account
- ระบบส่งอีเมลยืนยัน
- ระบบแสดงข้อความ "ลงทะเบียนสำเร็จ"
- Redirect ไปหน้า Login
```

**❌ Failure Path (1.1.5.2):**
```
- ระบบแสดงข้อความแจ้งเตือนเฉพาะเจาะจง:
  • "อีเมลนี้ถูกใช้แล้ว"
  • "รหัสผ่านไม่ปลอดภัย กรุณาใช้ตัวอักษรผสม"
  • "เบอร์โทรไม่ถูกต้อง กรุณากรอก 10 หลัก"
- กลับไปที่ขั้นตอน 1.1.3 (แก้ไขข้อมูล)
```

## 📝 **2.3 UCD-003: Booking Staff (ตัวอย่างเสริม)** [1 นาที]

### **Speaking Points:**
```
ตัวอย่างที่ 2 คือ UCD-003 Booking Staff
ซึ่งมี Validation ที่ซับซ้อนกว่า เพราะต้องตรวจสอบสิทธิ์และข้อมูลหลายชั้น
```

### **Key Validation Examples:**

#### **การสร้างการจอง (Create Booking):**
```
System Validations:
  ✓ ตรวจสอบสิทธิ์ของ Staff (role = "booking_staff")
  ✓ ตรวจสอบว่าลูกค้ามีอยู่ในระบบ
  ✓ ตรวจสอบว่าสนามมีอยู่จริง
  ✓ ตรวจสอบว่าวันเวลาที่เลือกอยู่ในอนาคต
  ✓ ตรวจสอบว่าสนามว่างในช่วงเวลานั้น
  ✓ ตรวจสอบว่าไม่มีการจองซ้อนทับ (Overlap)
  ✓ คำนวณราคาตามอัตราและระยะเวลา
```

#### **Success vs Failure:**
```
Success (ถ้าผ่านทุก Validation):
  → บันทึกการจอง
  → สร้างใบแจ้งหนี้
  → ส่งการแจ้งเตือนให้ลูกค้า
  → บันทึก Audit Log
  
Failure (ถ้ามี Validation ล้มเหลว):
  → แสดงข้อความเฉพาะเจาะจง
  → เช่น "สนามถูกจองแล้วในช่วงเวลา 14:00-16:00"
  → ให้เลือกเวลาใหม่
```

## 🔄 **3.1 Sequence Diagram: Login Flow** [1.5 นาที]

### **[แสดง Sequence Diagram - UCD-001 Login]**

### **Speaking Points:**
```
ตอนนี้เรามาดู Sequence Diagram ซึ่งแสดงการโต้ตอบระหว่าง Components
เริ่มจาก Login Flow ซึ่งเป็น Flow พื้นฐาน
```

---

### **📊 Key Components:**
1. **User** - ผู้ใช้งาน
2. **LoginController** - จัดการ login logic
3. **AccountDAO** - เข้าถึงข้อมูล account
4. **Database** - ฐานข้อมูล
5. **SessionManager** - จัดการ session
6. **AuditLog** - บันทึกการใช้งาน

---

### **🔄 Login Flow:**

```
User → LoginController: กรอก email + password
  |
  v
LoginController → AccountDAO: ตรวจสอบ credentials
  |
  v
AccountDAO → Database: SELECT * FROM Account WHERE email = ?
  |
  v
Database → AccountDAO: ส่งข้อมูล account กลับ
  |
  v
AccountDAO → LoginController: validate password hash
  |
  ├─ [SUCCESS PATH]
  |   LoginController → SessionManager: สร้าง session
  |   SessionManager → LoginController: session_id
  |   LoginController → AuditLog: บันทึก login success
  |   LoginController → User: redirect to dashboard
  |
  └─ [FAILURE PATH]
      LoginController → AccountDAO: increment failedLoginAttempts
      LoginController → AuditLog: บันทึก login failed
      LoginController → User: แสดง error message
```

---

### **🔐 Security Features:**
- Password hashing (BCrypt)
- Failed login attempts tracking (lock after 5 attempts)
- Session management
- Audit logging

## 📅 **3.2 Sequence Diagram: Booking Flow** [1.5 นาที]

### **[แสดง Sequence Diagram - UCD-003 Booking]**

### **Speaking Points:**
```
Booking Flow ซับซ้อนกว่า เพราะต้องตรวจสอบหลายอย่าง
และเกี่ยวข้องกับหลาย components
```

---

### **📊 Key Components:**
1. **BookingStaff** - พนักงานจอง
2. **BookingController** - จัดการ booking logic
3. **BookingDAO** - เข้าถึงข้อมูล booking
4. **FieldDAO** - เข้าถึงข้อมูลสนาม
5. **InvoiceController** - สร้างใบแจ้งหนี้
6. **NotificationService** - ส่งการแจ้งเตือน

---

### **🔄 Booking Flow:**

```
BookingStaff → BookingController: สร้างการจอง (fieldId, customerId, date, time)
  |
  v
BookingController → FieldDAO: ตรวจสอบว่าสนามมีอยู่
  |
  v
FieldDAO → Database: SELECT * FROM Field WHERE fieldId = ?
  |
  v
[CHECK 1: Field Exists?]
  |
  ├─ YES → BookingController → BookingDAO: ตรวจสอบว่าสนามว่าง
  |           |
  |           v
  |         BookingDAO → Database: SELECT * FROM Booking 
  |                                WHERE fieldId = ? AND date = ? 
  |                                AND timeSlot OVERLAPS ?
  |           |
  |           v
  |         [CHECK 2: Time Available?]
  |           |
  |           ├─ YES → BookingController → InvoiceController: สร้างใบแจ้งหนี้
  |           |           |
  |           |           v
  |           |         InvoiceController → Database: INSERT INTO Invoice
  |           |           |
  |           |           v
  |           |         BookingController → BookingDAO: บันทึกการจอง
  |           |           |
  |           |           v
  |           |         BookingDAO → Database: INSERT INTO Booking
  |           |           |
  |           |           v
  |           |         BookingController → NotificationService: ส่งแจ้งเตือน
  |           |           |
  |           |           v
  |           |         NotificationService → Customer: อีเมล/SMS แจ้งการจอง
  |           |           |
  |           |           v
  |           |         BookingController → BookingStaff: แสดงข้อความ "จองสำเร็จ"
  |           |
  |           └─ NO → BookingController → BookingStaff: "สนามถูกจองแล้ว"
  |
  └─ NO → BookingController → BookingStaff: "ไม่พบสนาม"
```

---

### **✅ Validation Points:**
1. ตรวจสอบว่า Field มีอยู่
2. ตรวจสอบว่าเวลาไม่ overlap
3. ตรวจสอบว่า Customer มีอยู่
4. สร้าง Invoice พร้อมกับ Booking (Transaction)
5. ส่งการแจ้งเตือนให้ลูกค้า

## 💳 **3.3 Sequence Diagram: Payment Processing** [1 นาที]

### **[แสดง Sequence Diagram - UCD-007 Payment]**

### **Speaking Points:**
```
สุดท้าย Payment Flow ที่เชื่อมต่อกับ External System
```

---

### **🔄 Payment Flow:**

```
Customer → PaymentController: เลือกวิธีชำระเงิน (PromptPay/Card)
  |
  v
PaymentController → InvoiceDAO: ดึงข้อมูล Invoice
  |
  v
InvoiceDAO → Database: SELECT * FROM Invoice WHERE invoiceId = ?
  |
  v
PaymentController → PaymentGateway: ส่งคำขอชำระเงิน
  |
  v
PaymentGateway → BankSystem: process payment
  |
  v
BankSystem → PaymentGateway: payment result
  |
  ├─ [SUCCESS]
  |   PaymentGateway → PaymentController: payment_success
  |   PaymentController → InvoiceDAO: อัพเดตสถานะ "PAID"
  |   PaymentController → BookingDAO: อัพเดต booking status "CONFIRMED"
  |   PaymentController → NotificationService: ส่งใบเสร็จ
  |   NotificationService → Customer: อีเมลใบเสร็จ
  |   PaymentController → Customer: แสดง "ชำระเงินสำเร็จ"
  |
  └─ [FAILURE]
      PaymentGateway → PaymentController: payment_failed
      PaymentController → Customer: แสดง "ชำระเงินไม่สำเร็จ กรุณาลองใหม่"
```

## 🏗️ **5.1 Email-based Architecture (จุดเด่นสำคัญ)** [1 นาที]

### **[แสดง Class Diagram]**

### **Speaking Points:**
```
ตอนนี้เรามาดู Class Diagram ซึ่งเป็นหัวใจสำคัญของระบบ
จุดเด่นที่สุดของ Design เราคือการใช้ Email เป็น Primary Key
แทนที่จะใช้ numeric ID แบบเดิม
```

---

### **📧 Email-based Primary Key Design:**

#### **เหตุผลที่เลือกใช้ Email:**
1. **User-Friendly:**
   - ผู้ใช้จำ email ได้ง่ายกว่า person_id
   - ไม่ต้องจำตัวเลข

2. **Authentication:**
   - ใช้ email ในการ Login
   - ใช้ email ในการ Forget Password
   - ใช้ email ในการสื่อสารกับลูกค้า

3. **Uniqueness:**
   - Email unique โดยธรรมชาติ
   - ไม่ซ้ำกันในโลกจริง

4. **Business Logic:**
   - รองรับการทำงานจริง
   - ลูกค้าใช้ email ในการติดต่อ

---

### **🔑 Primary Key Structure:**

```
Person (Abstract Class)
  - email {PK} VARCHAR(100)  ← Primary Key
  - firstName VARCHAR(100)
  - lastName VARCHAR(100)
  - phone VARCHAR(15)
  - address TEXT
```

**All Foreign Keys reference email:**
```
Account
  - accountId {PK} VARCHAR(50)
  - email {FK} → Person.email
  
Customer
  - email {PK,FK} → Person.email
  
Employee
  - email {PK,FK} → Person.email
  - employeeId VARCHAR(20) ← ไม่ใช่ PK แล้ว!
  
Booking
  - bookingId {PK}
  - customerEmail {FK} → Customer.email
  - approvedByEmail {FK} → BookingStaff.email
  
Invoice
  - invoiceId {PK}
  - customerEmail {FK} → Customer.email
  - createdByEmail {FK} → AccountingStaff.email
```

## 🏛️ **5.2 Inheritance Hierarchy** [1 นาที]

### **Speaking Points:**
```
ระบบเราใช้ Inheritance Pattern แบบ 3 ระดับ
เพื่อจัดการความสัมพันธ์ของผู้ใช้แต่ละประเภท
```

---

### **📊 Inheritance Structure:**

```
                    ┌─────────────────┐
                    │   Person        │ ← Abstract Class
                    │  (Abstract)     │
                    │                 │
                    │ - email {PK}    │
                    │ - firstName     │
                    │ - lastName      │
                    │ - phone         │
                    │ - address       │
                    └────────┬────────┘
                             │
                ┌────────────┴────────────┐
                │                         │
        ┌───────▼────────┐       ┌───────▼────────┐
        │   Customer     │       │   Employee     │ ← Abstract Class
        │                │       │  (Abstract)    │
        │                │       │                │
        │ LEVEL 2        │       │ - employeeId   │
        │ (Concrete)     │       │ - department   │
        │                │       │ - position     │
        │ - memberLevel  │       │ - salary       │
        │ - loyaltyPts   │       │ - hireDate     │
        └────────────────┘       └───────┬────────┘
                                         │
                    ┌────────────────────┼────────────────────┐
                    │                    │                    │
            ┌───────▼────────┐  ┌────────▼────────┐  ┌──────▼──────┐
            │ BookingStaff   │  │  SystemAdmin    │  │ Accounting  │
            │                │  │                 │  │    Staff    │
            │ LEVEL 3        │  │                 │  │             │
            │ (Concrete)     │  │ LEVEL 3         │  │ LEVEL 3     │
            │                │  │ (Concrete)      │  │ (Concrete)  │
            │ - specialty    │  │ - accessLevel   │  │ - license   │
            └────────────────┘  └─────────────────┘  └─────────────┘
            
            ┌────────────────┐  ┌─────────────────┐
            │ MarketingStaff │  │   Managing      │
            │                │  │   Director      │
            │ LEVEL 3        │  │                 │
            │ (Concrete)     │  │ LEVEL 3         │
            │                │  │ (Concrete)      │
            │ - campaigns    │  │ - department    │
            └────────────────┘  └─────────────────┘
```

---

### **🎯 Inheritance Benefits:**
1. **Code Reuse:** ข้อมูลพื้นฐานอยู่ใน Person
2. **Polymorphism:** ใช้ abstract methods (getRole(), getPermissions())
3. **Extensibility:** เพิ่ม Staff type ใหม่ได้ง่าย
4. **Type Safety:** แยกประเภทผู้ใช้ชัดเจน

### **7. Backup Q&A Answers:**

#### **Q: ทำไมใช้ email เป็น PK?**
```
A: เพราะ 3 เหตุผลหลัก
   1. User-friendly - ลูกค้าจำ email ได้ง่ายกว่า ID
   2. ใช้ในการ Login และ Forget Password อยู่แล้ว
   3. Unique โดยธรรมชาติ ไม่ต้อง Generate ID
```

#### **Q: Validation มีอะไรบ้าง?**
```
A: แบ่งเป็น 3 ระดับ
   1. Format Validation - อีเมล, เบอร์โทร, ฯลฯ
   2. Business Rule Validation - เวลาจองต้องในอนาคต, สนามต้องว่าง
   3. Permission Validation - ตรวจสอบสิทธิ์ตาม Role
```

#### **Q: รองรับการชำระเงินอะไรบ้าง?**
```
A: รองรับ 3 วิธี
   1. PromptPay - QR Code
   2. Credit Card - ผ่าน Gateway (Stripe/Omise)
   3. Bank Transfer - โอนเงินธนาคาร
```

#### **Q: ระบบมีความปลอดภัยอย่างไร?**
```
A: มี 4 มาตรการหลัก
   1. Password Hashing - BCrypt
   2. Failed Login Tracking - ล็อกหลัง 5 ครั้ง
   3. Session Management - Token-based
   4. Audit Logging - บันทึกทุก action
```
